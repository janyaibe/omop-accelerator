```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_silver.note (
  note_source_value STRING COMMENT 'Unique source key for note record',
  person_id BIGINT COMMENT 'FK -> person.person_id (OMOP: integer)',
  note_date DATE COMMENT 'Note date (OMOP: date)',
  note_datetime TIMESTAMP COMMENT 'Note datetime (OMOP: datetime)',
  note_type_concept_id INT COMMENT 'FK -> concept.concept_id (Type Concept) (OMOP: integer)',
  note_class_concept_id INT COMMENT 'FK -> concept.concept_id (Class Concept) (OMOP: integer)',
  note_title STRING COMMENT 'Note title (OMOP: varchar(250))',
  note_text STRING COMMENT 'Note text (OMOP: varchar(MAX))',
  encoding_concept_id INT COMMENT 'FK -> concept.concept_id (Encoding Concept) (OMOP: integer)',
  language_concept_id INT COMMENT 'FK -> concept.concept_id (Language Concept) (OMOP: integer)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_source_value STRING COMMENT 'Source visit occurrence key',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  source_system STRING COMMENT 'Source system (ex: allscripts_tw)',
  last_mod_tsp TIMESTAMP COMMENT 'Timestamp of last modification (OMOP: datetime)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 SILVER NOTE table';
```
