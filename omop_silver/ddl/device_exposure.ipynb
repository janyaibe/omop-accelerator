```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_silver.device_exposure (
  device_exposure_source_value STRING COMMENT 'Unique source key for device exposure record',
  person_id BIGINT COMMENT 'FK -> person.person_id (OMOP: integer)',
  device_concept_id INT COMMENT 'FK -> concept.concept_id (Standard Device) (OMOP: integer)',
  device_exposure_start_date DATE COMMENT 'Device exposure start date (OMOP: date)',
  device_exposure_start_datetime TIMESTAMP COMMENT 'Device exposure start datetime (OMOP: datetime)',
  device_exposure_end_date DATE COMMENT 'Device exposure end date (OMOP: date)',
  device_exposure_end_datetime TIMESTAMP COMMENT 'Device exposure end datetime (OMOP: datetime)',
  device_type_concept_id INT COMMENT 'FK -> concept.concept_id (Type Concept) (OMOP: integer)',
  unique_device_id STRING COMMENT 'Unique device identifier (OMOP: varchar(255))',
  production_id STRING COMMENT 'Production identifier / lot / serial (OMOP: varchar(255))',
  quantity DOUBLE COMMENT 'Quantity of device exposure (OMOP: float)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  device_source_value STRING COMMENT 'Source device code/value (OMOP: varchar(50))',
  device_source_concept_id INT COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
  unit_concept_id INT COMMENT 'FK -> concept.concept_id (Unit) (OMOP: integer)',
  unit_source_value STRING COMMENT 'Source unit value (OMOP: varchar(50))',
  unit_source_concept_id INT COMMENT 'FK -> concept.concept_id (Source unit concept) (OMOP: integer)',
  source_system STRING COMMENT 'Source system (ex: allscripts_tw)',
  last_mod_tsp TIMESTAMP COMMENT 'Timestamp of last modification (OMOP: datetime)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 SILVER DEVICE_EXPOSURE table';
```
