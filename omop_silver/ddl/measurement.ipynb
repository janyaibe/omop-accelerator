```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop_silver.measurement (
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  measurement_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Measurement) (OMOP: integer)',
  measurement_date DATE NOT NULL COMMENT 'Measurement date (OMOP: date)',
  measurement_datetime TIMESTAMP COMMENT 'Measurement datetime (optional) (OMOP: datetime)',
  measurement_time STRING COMMENT 'Measurement time (OMOP: varchar(10))',
  measurement_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Measurement type) (OMOP: integer)',
  operator_concept_id INT COMMENT 'FK -> concept.concept_id (Operator =, <, >, etc.) (OMOP: integer)',
  value_as_number DOUBLE COMMENT 'Numeric measurement value (OMOP: float)',
  value_as_concept_id INT COMMENT 'FK -> concept.concept_id (Value concept) (OMOP: integer)',
  unit_concept_id INT COMMENT 'FK -> concept.concept_id (Unit) (OMOP: integer)',
  range_low DOUBLE COMMENT 'Normal range lower bound (OMOP: float)',
  range_high DOUBLE COMMENT 'Normal range upper bound (OMOP: float)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  measurement_source_value STRING COMMENT 'Source measurement value (OMOP: varchar(50))',
  measurement_source_concept_id INT COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
  unit_source_value STRING COMMENT 'Source unit value (OMOP: varchar(50))',
  unit_source_concept_id INT COMMENT 'FK -> concept.concept_id (Unit source concept) (OMOP: integer)',
  value_source_value STRING COMMENT 'Source value as text (OMOP: varchar(50))',
  measurement_event_id BIGINT COMMENT 'FK to event table (OMOP: integer)',
  meas_event_field_concept_id INT COMMENT 'FK -> concept.concept_id (Event field) (OMOP: integer)',
  source_system STRING COMMENT 'Source system identifier',
  last_mod_tsp TIMESTAMP COMMENT 'Last modified timestamp'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 MEASUREMENT silver layer';
```
