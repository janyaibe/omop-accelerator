```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_silver.procedure_occurrence (
  procedure_occurrence_source_value STRING COMMENT 'Unique source key for procedure record',
  person_id BIGINT COMMENT 'FK -> person.person_id (OMOP: integer)',
  procedure_concept_id INT COMMENT 'FK -> concept.concept_id (Standard Procedure) (OMOP: integer)',
  procedure_date DATE COMMENT 'Procedure date (OMOP: date)',
  procedure_datetime TIMESTAMP COMMENT 'Procedure datetime (OMOP: datetime)',
  procedure_type_concept_id INT COMMENT 'FK -> concept.concept_id (Type Concept) (OMOP: integer)',
  modifier_concept_id INT COMMENT 'FK -> concept.concept_id (Modifier) (OMOP: integer)',
  quantity DOUBLE COMMENT 'Quantity of procedure (OMOP: float)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  procedure_source_value STRING COMMENT 'Source procedure code (OMOP: varchar(50))',
  procedure_source_concept_id INT COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
  modifier_source_value STRING COMMENT 'Source modifier code (OMOP: varchar(50))',
  source_system STRING COMMENT 'Source system (ex: allscripts_tw)',
  last_mod_tsp TIMESTAMP COMMENT 'Timestamp of last modification (OMOP: datetime)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 SILVER PROCEDURE_OCCURRENCE table';
```
