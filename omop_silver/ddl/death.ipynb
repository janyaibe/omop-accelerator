```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_silver.death (
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  death_date DATE NOT NULL COMMENT 'Date of death (OMOP: date)',
  death_datetime TIMESTAMP COMMENT 'Datetime of death (optional) (OMOP: datetime)',
  death_type_concept_id INT COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  cause_concept_id INT COMMENT 'FK -> concept.concept_id (Cause) (OMOP: integer)',
  cause_source_value STRING COMMENT 'Source cause (OMOP: varchar(50))',
  cause_source_concept_id INT COMMENT 'FK -> concept.concept_id (Cause source concept) (OMOP: integer)',
  source_system STRING NOT NULL COMMENT 'Source system identifier (e.g., allscripts_tw)',
  last_mod_tsp TIMESTAMP NOT NULL COMMENT 'Last modified timestamp in silver'

)
USING DELTA
COMMENT 'Silver staging table for OMOP CDM v5.4 DEATH';
```
