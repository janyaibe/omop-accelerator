```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_silver.observation (
  observation_source_value STRING COMMENT 'Unique source key for observation record',

  person_id BIGINT COMMENT 'FK -> person.person_id (OMOP: integer)',
  observation_concept_id INT COMMENT 'FK -> concept.concept_id (Observation) (OMOP: integer)',

  observation_date DATE COMMENT 'Observation date (OMOP: date)',
  observation_datetime TIMESTAMP COMMENT 'Observation datetime (OMOP: datetime)',

  observation_type_concept_id INT COMMENT 'FK -> concept.concept_id (Observation type) (OMOP: integer)',

  value_as_number DOUBLE COMMENT 'Numeric observation value (OMOP: float)',
  value_as_string STRING COMMENT 'Text observation value (OMOP: varchar(60))',
  value_as_concept_id INT COMMENT 'FK -> concept.concept_id (Value concept) (OMOP: integer)',

  qualifier_concept_id INT COMMENT 'FK -> concept.concept_id (Qualifier) (OMOP: integer)',
  unit_concept_id INT COMMENT 'FK -> concept.concept_id (Unit) (OMOP: integer)',

  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',

  observation_source_code STRING COMMENT 'Original source observation value',
  observation_source_concept_id INT COMMENT 'FK -> concept.concept_id (Source concept)',

  unit_source_value STRING COMMENT 'Source unit value',
  qualifier_source_value STRING COMMENT 'Source qualifier value',
  value_source_value STRING COMMENT 'Source value',

  observation_event_id BIGINT COMMENT 'FK to event table (OMOP: integer)',
  obs_event_field_concept_id INT COMMENT 'FK -> concept.concept_id (Event field)',

  source_system STRING COMMENT 'Source system identifier',
  last_mod_tsp TIMESTAMP COMMENT 'Last modified timestamp'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 OBSERVATION silver layer';
```
