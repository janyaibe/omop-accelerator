```python
mapping_tables = [
    "source_to_person",
    "source_to_location",
    "source_to_care_site",
    "source_to_provider",
    "source_to_visit_occurrence",
    "source_to_visit_detail",
    "source_to_condition_occurrence",
    "source_to_procedure_occurrence",
    "source_to_measurement",
    "source_to_observation",
    "source_to_drug_exposure",
    "source_to_device_exposure"]

omop_tables = [
    "person",
    "location",
    "care_site",
    "provider",
    "visit_occurrence",
    "visit_detail",
    "condition_occurrence",
    "procedure_occurrence",
    "measurement",
    "observation",
    "drug_exposure",
    "device_exposure",
    "death",
    "note"
]

vocab_tables = [
    "concept",
    "concept_ancestor",
    "concept_class",
    "concept_relationship",
    "concept_synonym",
    "domain",
    "relationship",
    "vocabulary"
]
```

```python
catalog = "_exponent"
schema = "omop_mapping"
source_filter = "allscripts%"

for table in mapping_tables:
    full_table = f"{catalog}.{schema}.{table}"
    
    try:
        spark.sql(f"""
            DELETE FROM {full_table}
            WHERE source_system LIKE '{source_filter}'
        """)
        print(f"{full_table}")
    except Exception as e:
        print(f"{full_table} — {str(e)}")
```

```python
catalog = "_exponent"
schema = "omop_silver"
source_filter = "allscripts%"

for table in omop_tables:
    full_table = f"{catalog}.{schema}.{table}"
    
    sql = f"""
    DELETE FROM {full_table}
    WHERE source_system LIKE '{source_filter}'
    """
    
    print(sql.strip())
    
    try:
        spark.sql(sql)
        print(f"Success: {full_table}")
    except Exception as e:
        print(f"Failed: {full_table}")
        print(f"Error: {str(e)}")
```
