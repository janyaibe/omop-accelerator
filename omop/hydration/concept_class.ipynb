```python
from pyspark.sql.functions import col
```

```python
concept_class_df = (
    spark.read
        .option("header", True)
        .option("sep", "\t")
        .csv("/Volumes/_exponent/omop_files/vocabulary/CONCEPT_CLASS.csv")
        .withColumn("concept_class_concept_id", col("concept_class_concept_id").cast("int"))
)

concept_class_df.write.format("delta").mode("overwrite").saveAsTable("_exponent.omop.concept_class")
```
