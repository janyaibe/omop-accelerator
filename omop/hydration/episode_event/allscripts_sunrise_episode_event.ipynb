# Allscripts SCM Episode Event Hydration

Link SCM episodes to related conditions, visits, measurements, and drug exposures.

## Table DDL
`event_id` stores the **domain row id** (e.g. `condition_occurrence_id`). If `omop_scm.episode_event` was created with `event_id` as `GENERATED ALWAYS AS IDENTITY`, inserts fail — the first SQL cell recreates the table with OMOP-correct columns.

## Condition linkage
The **condition_occurrence** branch matches rows the same way as `allscripts_scm_episode` rollup when possible: **`concept_ancestor`** (canonical + **swapped-column**), **exact** `condition_concept_id = episode_object_concept_id`, and **`Is a`** chains (up to 4 hops) via `isa_condition`. Without this, SCM often gets **zero** condition links because `condition_concept_id` never appears as `descendant_concept_id` in `concept_ancestor`.

Run **`TRUNCATE episode_event`** in this notebook before loading, or rely on `allscripts_scm_episode` (it truncates `episode_event` first) then run this notebook.

## Episode event field concepts
- 1147127: `condition_occurrence.condition_occurrence_id`
- 1147126: `visit_occurrence.visit_occurrence_id`
- 1147130: `measurement.measurement_id`
- 1147132: `drug_exposure.drug_exposure_id`


```python
%sql
TRUNCATE TABLE _exponent.omop_scm.episode_event;
```

```python
%sql
-- OMOP: event_id is the FK to the domain table row, not an identity column.
CREATE OR REPLACE TABLE _exponent.omop_scm.episode_event (
  episode_event_id BIGINT GENERATED BY DEFAULT AS IDENTITY NOT NULL,
  episode_id BIGINT NOT NULL,
  event_id BIGINT NOT NULL,
  episode_event_field_concept_id INT NOT NULL,
  episode_event_concept_id INT
)
USING DELTA
COMMENT 'OMOP CDM v5.4 EPISODE_EVENT (SCM)';
```

```python
%sql
CREATE OR REPLACE TEMP VIEW isa_condition AS
SELECT
  CAST(concept_id_1 AS BIGINT) AS child_concept_id,
  CAST(concept_id_2 AS BIGINT) AS parent_concept_id
FROM _exponent.omop.concept_relationship
WHERE relationship_id = 'Is a'
  AND invalid_reason IS NULL;

CREATE OR REPLACE TEMP VIEW episode_event_candidate AS

SELECT
  e.episode_id,
  co.condition_occurrence_id AS event_id,
  1147127 AS episode_event_field_concept_id
FROM _exponent.omop_scm.episode e
JOIN _exponent.omop_scm.condition_occurrence co
  ON co.person_id = e.person_id
 AND co.condition_start_date >= e.episode_start_date
 AND co.condition_concept_id <> 0
JOIN _exponent.omop.concept_ancestor ca
  ON CAST(ca.descendant_concept_id AS BIGINT) = CAST(co.condition_concept_id AS BIGINT)
 AND CAST(ca.ancestor_concept_id AS BIGINT) = CAST(e.episode_object_concept_id AS BIGINT)

UNION ALL

SELECT
  e.episode_id,
  co.condition_occurrence_id AS event_id,
  1147127 AS episode_event_field_concept_id
FROM _exponent.omop_scm.episode e
JOIN _exponent.omop_scm.condition_occurrence co
  ON co.person_id = e.person_id
 AND co.condition_start_date >= e.episode_start_date
 AND co.condition_concept_id <> 0
JOIN _exponent.omop.concept_ancestor ca
  ON CAST(ca.ancestor_concept_id AS BIGINT) = CAST(co.condition_concept_id AS BIGINT)
 AND CAST(ca.descendant_concept_id AS BIGINT) = CAST(e.episode_object_concept_id AS BIGINT)

UNION ALL

SELECT
  e.episode_id,
  co.condition_occurrence_id AS event_id,
  1147127 AS episode_event_field_concept_id
FROM _exponent.omop_scm.episode e
JOIN _exponent.omop_scm.condition_occurrence co
  ON co.person_id = e.person_id
 AND co.condition_start_date >= e.episode_start_date
 AND co.condition_concept_id <> 0
 AND CAST(co.condition_concept_id AS BIGINT) = CAST(e.episode_object_concept_id AS BIGINT)

UNION ALL

SELECT
  e.episode_id,
  co.condition_occurrence_id AS event_id,
  1147127 AS episode_event_field_concept_id
FROM _exponent.omop_scm.episode e
JOIN _exponent.omop_scm.condition_occurrence co
  ON co.person_id = e.person_id
 AND co.condition_start_date >= e.episode_start_date
 AND co.condition_concept_id <> 0
JOIN isa_condition i1
  ON i1.child_concept_id = CAST(co.condition_concept_id AS BIGINT)
 AND i1.parent_concept_id = CAST(e.episode_object_concept_id AS BIGINT)

UNION ALL

SELECT
  e.episode_id,
  co.condition_occurrence_id AS event_id,
  1147127 AS episode_event_field_concept_id
FROM _exponent.omop_scm.episode e
JOIN _exponent.omop_scm.condition_occurrence co
  ON co.person_id = e.person_id
 AND co.condition_start_date >= e.episode_start_date
 AND co.condition_concept_id <> 0
JOIN isa_condition i1
  ON i1.child_concept_id = CAST(co.condition_concept_id AS BIGINT)
JOIN isa_condition i2
  ON i2.child_concept_id = i1.parent_concept_id
 AND i2.parent_concept_id = CAST(e.episode_object_concept_id AS BIGINT)

UNION ALL

SELECT
  e.episode_id,
  co.condition_occurrence_id AS event_id,
  1147127 AS episode_event_field_concept_id
FROM _exponent.omop_scm.episode e
JOIN _exponent.omop_scm.condition_occurrence co
  ON co.person_id = e.person_id
 AND co.condition_start_date >= e.episode_start_date
 AND co.condition_concept_id <> 0
JOIN isa_condition i1
  ON i1.child_concept_id = CAST(co.condition_concept_id AS BIGINT)
JOIN isa_condition i2
  ON i2.child_concept_id = i1.parent_concept_id
JOIN isa_condition i3
  ON i3.child_concept_id = i2.parent_concept_id
 AND i3.parent_concept_id = CAST(e.episode_object_concept_id AS BIGINT)

UNION ALL

SELECT
  e.episode_id,
  co.condition_occurrence_id AS event_id,
  1147127 AS episode_event_field_concept_id
FROM _exponent.omop_scm.episode e
JOIN _exponent.omop_scm.condition_occurrence co
  ON co.person_id = e.person_id
 AND co.condition_start_date >= e.episode_start_date
 AND co.condition_concept_id <> 0
JOIN isa_condition i1
  ON i1.child_concept_id = CAST(co.condition_concept_id AS BIGINT)
JOIN isa_condition i2
  ON i2.child_concept_id = i1.parent_concept_id
JOIN isa_condition i3
  ON i3.child_concept_id = i2.parent_concept_id
JOIN isa_condition i4
  ON i4.child_concept_id = i3.parent_concept_id
 AND i4.parent_concept_id = CAST(e.episode_object_concept_id AS BIGINT)

UNION ALL

SELECT
  e.episode_id,
  vo.visit_occurrence_id AS event_id,
  1147126 AS episode_event_field_concept_id
FROM _exponent.omop_scm.episode e
JOIN _exponent.omop_scm.visit_occurrence vo
  ON vo.person_id = e.person_id
WHERE vo.visit_start_date >= e.episode_start_date

UNION ALL

SELECT
  e.episode_id,
  m.measurement_id AS event_id,
  1147130 AS episode_event_field_concept_id
FROM _exponent.omop_scm.episode e
JOIN _exponent.omop_scm.measurement m
  ON m.person_id = e.person_id
WHERE m.measurement_date >= e.episode_start_date

UNION ALL

SELECT
  e.episode_id,
  de.drug_exposure_id AS event_id,
  1147132 AS episode_event_field_concept_id
FROM _exponent.omop_scm.episode e
JOIN _exponent.omop_scm.drug_exposure de
  ON de.person_id = e.person_id
WHERE de.drug_exposure_start_date >= e.episode_start_date;
```

```python
%sql
INSERT INTO _exponent.omop_scm.episode_event (
  episode_id,
  event_id,
  episode_event_field_concept_id
)
SELECT DISTINCT
  episode_id,
  event_id,
  episode_event_field_concept_id
FROM episode_event_candidate;
```

```python
%sql
SELECT
  episode_event_field_concept_id,
  CASE episode_event_field_concept_id
    WHEN 1147127 THEN 'condition_occurrence'
    WHEN 1147126 THEN 'visit_occurrence'
    WHEN 1147130 THEN 'measurement'
    WHEN 1147132 THEN 'drug_exposure'
    ELSE 'other'
  END AS event_type,
  COUNT(*) AS event_count
FROM _exponent.omop_scm.episode_event
GROUP BY episode_event_field_concept_id
ORDER BY event_count DESC;
```
