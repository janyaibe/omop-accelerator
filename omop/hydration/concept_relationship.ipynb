```python
from pyspark.sql.functions import col, to_date
```

```python
concept_relationship_df = (
    spark.read
        .option("header", True)
        .option("sep", "\t")
        .csv("/Volumes/_exponent/omop_files/vocabulary/CONCEPT_RELATIONSHIP.csv")
        .withColumn("concept_id_1", col("concept_id_1").cast("int"))
        .withColumn("concept_id_2", col("concept_id_2").cast("int"))
        .withColumn("valid_start_date", to_date(col("valid_start_date"), "yyyyMMdd"))
        .withColumn("valid_end_date", to_date(col("valid_end_date"), "yyyyMMdd"))
)

concept_relationship_df.write.format("delta").mode("overwrite").saveAsTable("_exponent.omop.concept_relationship")
```
