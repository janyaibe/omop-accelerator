```python
from pyspark.sql.functions import col
```

```python
domain_df = (
    spark.read
        .option("header", True)
        .option("sep", "\t")
        .csv("/Volumes/_exponent/omop_files/vocabulary/DOMAIN.csv")
        .withColumn("domain_concept_id", col("domain_concept_id").cast("int"))

)

domain_df.write.format("delta").mode("overwrite").saveAsTable("_exponent.omop.domain")
```
