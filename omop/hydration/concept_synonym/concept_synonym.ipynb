```python
from pyspark.sql.functions import col
```

```python
concept_synonym_df = (
    spark.read
        .option("header", True)
        .option("sep", "\t")
        .csv("/Volumes/_exponent/omop_files/vocabulary/CONCEPT_SYNONYM.csv")
        .withColumn("concept_id", col("concept_id").cast("int"))
        .withColumn("language_concept_id", col("language_concept_id").cast("int"))

)

concept_synonym_df.write.format("delta").mode("overwrite").saveAsTable("_exponent.omop.concept_synonym")
```
