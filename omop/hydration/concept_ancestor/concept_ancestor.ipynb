```python
from pyspark.sql import functions as F

concept_ancestor_df = (
    spark.read
    .option("header", True)
    .option("sep", "\t")
    .csv("/Volumes/_exponent/omop_files/vocabulary/CONCEPT_ANCESTOR.csv")
    .select(
        F.col("ancestor_concept_id").cast("int").alias("ancestor_concept_id"),
        F.col("descendant_concept_id").cast("int").alias("descendant_concept_id"),
        F.col("min_levels_of_separation").cast("int").alias("min_levels_of_separation"),
        F.col("max_levels_of_separation").cast("int").alias("max_levels_of_separation"),
    )
)

concept_ancestor_df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("_exponent.omop_allscripts.concept_ancestor")
concept_ancestor_df.write.mode("overwrite").option("overwriteSchema", "true").saveAsTable("_exponent.omop_epic.concept_ancestor")
concept_ancestor_df.display()
```
