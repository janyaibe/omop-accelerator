```python
from pyspark.sql.functions import col, to_date
```

```python
drug_strength_df = (
    spark.read
        .option("header", True)
        .option("sep", "\t")
        .csv("/Volumes/_exponent/omop_files/vocabulary/DRUG_STRENGTH.csv")
        .select(
            col("drug_concept_id").cast("int").alias("drug_concept_id"),
            col("ingredient_concept_id").cast("int").alias("ingredient_concept_id"),
            col("amount_value").cast("double").alias("amount_value"),
            col("amount_unit_concept_id").cast("int").alias("amount_unit_concept_id"),
            col("numerator_value").cast("double").alias("numerator_value"),
            col("numerator_unit_concept_id").cast("int").alias("numerator_unit_concept_id"),
            col("denominator_value").cast("double").alias("denominator_value"),
            col("denominator_unit_concept_id").cast("int").alias("denominator_unit_concept_id"),
            col("box_size").cast("int").alias("box_size"),
            to_date(col("valid_start_date"), "yyyyMMdd").alias("valid_start_date"),
            to_date(col("valid_end_date"), "yyyyMMdd").alias("valid_end_date"),
            col("invalid_reason").cast("string").alias("invalid_reason"),
        )
)

drug_strength_df.write.format("delta").mode("overwrite").saveAsTable("_exponent.omop.drug_strength")
```
