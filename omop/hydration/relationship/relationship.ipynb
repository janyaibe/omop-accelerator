```python
from pyspark.sql.functions import col

relationship_df = (
    spark.read
        .option("header", True)
        .option("sep", "\t")
        .csv("/Volumes/_exponent/omop_files/vocabulary/RELATIONSHIP.csv")
        .select(
            col("relationship_id").cast("string").alias("relationship_id"),
            col("relationship_name").cast("string").alias("relationship_name"),
            col("is_hierarchical").cast("string").alias("is_hierarchical"),
            col("defines_ancestry").cast("string").alias("defines_ancestry"),
            col("reverse_relationship_id").cast("string").alias("reverse_relationship_id"),
            col("relationship_concept_id").cast("int").alias("relationship_concept_id"),
        )
)

relationship_df.write.format("delta").mode("overwrite").saveAsTable("_exponent.omop.relationship")
```
