```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.concept_synonym (
  concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (OMOP: integer)',
  concept_synonym_name STRING NOT NULL COMMENT 'Synonym text (OMOP: varchar(1000))',
  language_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Language) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 CONCEPT_SYNONYM table';
```
