```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.relationship (
  relationship_id STRING NOT NULL COMMENT 'Relationship identifier (OMOP: varchar(20))',
  relationship_name STRING NOT NULL COMMENT 'Relationship name (OMOP: varchar(255))',
  is_hierarchical STRING NOT NULL COMMENT 'Y/N (OMOP: varchar(1))',
  defines_ancestry STRING NOT NULL COMMENT 'Y/N (OMOP: varchar(1))',
  reverse_relationship_id STRING NOT NULL COMMENT 'FK -> relationship.relationship_id (OMOP: varchar(20))',
  relationship_concept_id INT NOT NULL COMMENT 'Concept representing relationship (FK -> concept.concept_id) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 RELATIONSHIP table';
```
