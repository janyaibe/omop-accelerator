```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.metadata (
  metadata_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Metadata concept) (OMOP: integer)',
  metadata_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Metadata type) (OMOP: integer)',
  name STRING NOT NULL COMMENT 'Metadata name (OMOP: varchar(250))',
  value_as_string STRING COMMENT 'Value as string (optional) (OMOP: varchar(250))',
  value_as_concept_id INT COMMENT 'FK -> concept.concept_id (Value as concept, optional) (OMOP: integer)',
  metadata_date DATE COMMENT 'Metadata date (optional) (OMOP: date)',
  metadata_datetime TIMESTAMP COMMENT 'Metadata datetime (optional) (OMOP: datetime)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 METADATA table';
```
