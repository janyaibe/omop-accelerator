```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.attribute_definition (
  attribute_definition_id INT NOT NULL COMMENT 'PK (OMOP: integer)',
  attribute_name STRING NOT NULL COMMENT 'Name (OMOP: varchar(255))',
  attribute_description STRING COMMENT 'Description (optional) (OMOP: varchar(1000))',
  attribute_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  attribute_syntax STRING COMMENT 'Syntax/definition (optional) (OMOP: varchar(MAX))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 ATTRIBUTE_DEFINITION table';
```
