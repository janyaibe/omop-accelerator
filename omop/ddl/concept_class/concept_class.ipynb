```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.concept_class (
  concept_class_id STRING NOT NULL COMMENT 'Concept class identifier (OMOP: varchar(20))',
  concept_class_name STRING NOT NULL COMMENT 'Concept class name (OMOP: varchar(255))',
  concept_class_concept_id INT NOT NULL COMMENT 'Concept representing the concept class (FK -> concept.concept_id) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 CONCEPT_CLASS table';
```
