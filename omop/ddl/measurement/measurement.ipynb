```python
%sql
CREATE OR REPLACE TABLE _exponent.omop.measurement (
  measurement_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  measurement_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Measurement)',
  measurement_date DATE NOT NULL COMMENT 'Measurement date (OMOP: date)',
  measurement_datetime TIMESTAMP COMMENT 'Measurement datetime (optional)',
  measurement_time STRING COMMENT 'Time (optional) (OMOP: varchar(10))',
  measurement_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type)',
  operator_concept_id INT COMMENT 'FK -> concept.concept_id (Operator)',
  value_as_number DOUBLE COMMENT 'Numeric value',
  value_as_concept_id INT COMMENT 'FK -> concept.concept_id (Coded value)',
  unit_concept_id INT COMMENT 'FK -> concept.concept_id (Unit)',
  range_low DOUBLE COMMENT 'Reference range low',
  range_high DOUBLE COMMENT 'Reference range high',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id',
  measurement_source_value STRING COMMENT 'Source measurement value',
  measurement_source_concept_id INT COMMENT 'FK -> concept.concept_id (Source concept)',
  unit_source_value STRING COMMENT 'Source unit',
  unit_source_concept_id INT COMMENT 'FK -> concept.concept_id (Source unit concept)',
  value_source_value STRING COMMENT 'Source value',
  measurement_event_id BIGINT COMMENT 'Measurement event identifier',
  meas_event_field_concept_id INT COMMENT 'Measurement event field concept id'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 MEASUREMENT table';
```
