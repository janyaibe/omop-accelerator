```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.measurement (
  measurement_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  measurement_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Measurement) (OMOP: integer)',
  measurement_date DATE NOT NULL COMMENT 'Measurement date (OMOP: date)',
  measurement_datetime TIMESTAMP COMMENT 'Measurement datetime (optional) (OMOP: datetime)',
  measurement_time STRING COMMENT 'Time (optional) (OMOP: varchar(10))',
  measurement_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  operator_concept_id INT COMMENT 'FK -> concept.concept_id (Operator) (OMOP: integer)',
  value_as_number DOUBLE COMMENT 'Numeric value (optional) (OMOP: numeric)',
  value_as_concept_id INT COMMENT 'FK -> concept.concept_id (Coded value) (OMOP: integer)',
  unit_concept_id INT COMMENT 'FK -> concept.concept_id (Unit) (OMOP: integer)',
  range_low DOUBLE COMMENT 'Reference range low (optional) (OMOP: numeric)',
  range_high DOUBLE COMMENT 'Reference range high (optional) (OMOP: numeric)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  measurement_source_value STRING COMMENT 'Source measurement value (OMOP: varchar(50))',
  measurement_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
  unit_source_value STRING COMMENT 'Source unit (OMOP: varchar(50))',
  value_source_value STRING COMMENT 'Source value (optional) (OMOP: varchar(50))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 MEASUREMENT table';
```
