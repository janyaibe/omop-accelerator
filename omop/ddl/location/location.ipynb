```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.location (
  location_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  address_1 STRING COMMENT 'Address line 1 (OMOP: varchar(50))',
  address_2 STRING COMMENT 'Address line 2 (OMOP: varchar(50))',
  city STRING COMMENT 'City (OMOP: varchar(50))',
  state STRING COMMENT 'State (OMOP: varchar(2))',
  zip STRING COMMENT 'ZIP (OMOP: varchar(9))',
  county STRING COMMENT 'County (OMOP: varchar(20))',
  location_source_value STRING COMMENT 'Source location value (OMOP: varchar(50))',
  country_concept_id INT COMMENT 'FK -> concept.concept_id (Country) (OMOP: integer)',
  country_source_value STRING COMMENT 'Source country (OMOP: varchar(80))',
  latitude DOUBLE COMMENT 'Latitude (optional) (OMOP: float)',
  longitude DOUBLE COMMENT 'Longitude (optional) (OMOP: float)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 LOCATION table';
```
