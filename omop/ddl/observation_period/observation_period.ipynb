```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.observation_period (
  observation_period_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  observation_period_start_date DATE NOT NULL COMMENT 'Start date of observation period (OMOP: date)',
  observation_period_end_date DATE NOT NULL COMMENT 'End date of observation period (OMOP: date)',
  period_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type of observation period) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 OBSERVATION_PERIOD table';
```
