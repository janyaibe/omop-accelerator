```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.drug_exposure (
  drug_exposure_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  drug_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Drug) (OMOP: integer)',
  drug_exposure_start_date DATE NOT NULL COMMENT 'Start date (OMOP: date)',
  drug_exposure_start_datetime TIMESTAMP COMMENT 'Start datetime (optional) (OMOP: datetime)',
  drug_exposure_end_date DATE NOT NULL COMMENT 'End date (OMOP: date)',
  drug_exposure_end_datetime TIMESTAMP COMMENT 'End datetime (optional) (OMOP: datetime)',
  verbatim_end_date DATE COMMENT 'Source end date (optional) (OMOP: date)',
  drug_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  stop_reason STRING COMMENT 'Stop reason (optional) (OMOP: varchar(20))',
  refills INT COMMENT 'Refills (optional) (OMOP: integer)',
  quantity DOUBLE COMMENT 'Quantity (optional) (OMOP: numeric)',
  days_supply INT COMMENT 'Days supply (optional) (OMOP: integer)',
  sig STRING COMMENT 'Instructions (optional) (OMOP: varchar(MAX))',
  route_concept_id INT COMMENT 'FK -> concept.concept_id (Route) (OMOP: integer)',
  lot_number STRING COMMENT 'Lot number (optional) (OMOP: varchar(50))',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  drug_source_value STRING COMMENT 'Source drug value (OMOP: varchar(50))',
  drug_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
  route_source_value STRING COMMENT 'Source route (OMOP: varchar(50))',
  dose_unit_source_value STRING COMMENT 'Source dose unit (OMOP: varchar(50))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 DRUG_EXPOSURE table';
```
