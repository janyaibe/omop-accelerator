```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.note (
  note_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  note_date DATE NOT NULL COMMENT 'Note date (OMOP: date)',
  note_datetime TIMESTAMP COMMENT 'Note datetime (optional) (OMOP: datetime)',
  note_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  note_class_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Class) (OMOP: integer)',
  note_title STRING COMMENT 'Title (optional) (OMOP: varchar(250))',
  note_text STRING NOT NULL COMMENT 'Full note text (OMOP: varchar(MAX))',
  encoding_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Encoding) (OMOP: integer)',
  language_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Language) (OMOP: integer)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  note_source_value STRING COMMENT 'Source note type/value (OMOP: varchar(50))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 NOTE table';
```
