```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.care_site (
  care_site_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  care_site_name STRING COMMENT 'Care site name (OMOP: varchar(255))',
  place_of_service_concept_id INT COMMENT 'FK -> concept.concept_id (Place of service) (OMOP: integer)',
  location_id BIGINT COMMENT 'FK -> location.location_id (OMOP: integer)',
  care_site_source_value STRING COMMENT 'Source care site value (OMOP: varchar(50))',
  place_of_service_source_value STRING COMMENT 'Source place of service (OMOP: varchar(50))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 CARE_SITE table';
```
