```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.visit_detail (
  visit_detail_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  visit_detail_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Visit detail) (OMOP: integer)',
  visit_detail_start_date DATE NOT NULL COMMENT 'Start date (OMOP: date)',
  visit_detail_start_datetime TIMESTAMP COMMENT 'Start datetime (optional) (OMOP: datetime)',
  visit_detail_end_date DATE NOT NULL COMMENT 'End date (OMOP: date)',
  visit_detail_end_datetime TIMESTAMP COMMENT 'End datetime (optional) (OMOP: datetime)',
  visit_detail_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  care_site_id BIGINT COMMENT 'FK -> care_site.care_site_id (OMOP: integer)',
  visit_detail_source_value STRING COMMENT 'Source value (OMOP: varchar(50))',
  visit_detail_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
  admitted_from_concept_id INT COMMENT 'FK -> concept.concept_id (OMOP: integer)',
  admitted_from_source_value STRING COMMENT 'Source admit value (OMOP: varchar(50))',
  discharged_to_concept_id INT COMMENT 'FK -> concept.concept_id (OMOP: integer)',
  discharged_to_source_value STRING COMMENT 'Source discharge value (OMOP: varchar(50))',
  preceding_visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  parent_visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  visit_occurrence_id BIGINT NOT NULL COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 VISIT_DETAIL table';
```
