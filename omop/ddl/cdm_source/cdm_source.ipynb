```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.cdm_source (
  cdm_source_name STRING NOT NULL COMMENT 'Name of source (OMOP: varchar(255))',
  cdm_source_abbreviation STRING NOT NULL COMMENT 'Abbrev (OMOP: varchar(25))',
  cdm_holder STRING NOT NULL COMMENT 'Holder (OMOP: varchar(255))',
  source_description STRING COMMENT 'Description (OMOP: varchar(MAX))',
  source_documentation_reference STRING COMMENT 'Documentation reference (OMOP: varchar(255))',
  cdm_etl_reference STRING COMMENT 'ETL reference (OMOP: varchar(255))',
  source_release_date DATE NOT NULL COMMENT 'Source release date (OMOP: date)',
  cdm_release_date DATE NOT NULL COMMENT 'CDM release date (OMOP: date)',
  cdm_version STRING COMMENT 'CDM version (e.g., v5.4) (OMOP: varchar(10))',
  cdm_version_concept_id INT NOT NULL COMMENT 'Concept Id representing CDM version (FK -> concept.concept_id) (OMOP: integer)',
  vocabulary_version STRING NOT NULL COMMENT 'Vocabulary version string (OMOP: varchar(20))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 CDM_SOURCE table';
```
