```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.concept_relationship (
  concept_id_1 INT NOT NULL COMMENT 'FK -> concept.concept_id (OMOP: integer)',
  concept_id_2 INT NOT NULL COMMENT 'FK -> concept.concept_id (OMOP: integer)',
  relationship_id STRING NOT NULL COMMENT 'FK -> relationship.relationship_id (OMOP: varchar(20))',
  valid_start_date DATE NOT NULL COMMENT 'Valid start (OMOP: date)',
  valid_end_date DATE NOT NULL COMMENT 'Valid end (OMOP: date)',
  invalid_reason STRING COMMENT 'Invalid reason (optional) (OMOP: varchar(1))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 CONCEPT_RELATIONSHIP table';
```
