```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.note_nlp (
  note_nlp_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  note_id BIGINT NOT NULL COMMENT 'FK -> note.note_id (OMOP: integer)',
  section_concept_id INT COMMENT 'FK -> concept.concept_id (Section) (OMOP: integer)',
  snippet STRING COMMENT 'Snippet text (optional) (OMOP: varchar(250))',
  offset STRING COMMENT 'Character offsets (optional) (OMOP: varchar(250))',
  lexical_variant STRING NOT NULL COMMENT 'Lexical variant (OMOP: varchar(250))',
  note_nlp_concept_id INT COMMENT 'FK -> concept.concept_id (NLP concept) (OMOP: integer)',
  note_nlp_source_concept_id INT COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
  nlp_system STRING NOT NULL COMMENT 'NLP system name (OMOP: varchar(250))',
  nlp_date DATE NOT NULL COMMENT 'NLP run date (OMOP: date)',
  nlp_datetime TIMESTAMP COMMENT 'NLP run datetime (optional) (OMOP: datetime)',
  term_exists STRING COMMENT 'Y/N (optional) (OMOP: varchar(1))',
  term_temporal STRING COMMENT 'Temporal (optional) (OMOP: varchar(50))',
  term_modifiers STRING COMMENT 'Modifiers (optional) (OMOP: varchar(2000))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 NOTE_NLP table';
```
