```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.episode (
  episode_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  episode_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Episode) (OMOP: integer)',
  episode_start_date DATE NOT NULL COMMENT 'Start date (OMOP: date)',
  episode_start_datetime TIMESTAMP COMMENT 'Start datetime (optional) (OMOP: datetime)',
  episode_end_date DATE COMMENT 'End date (optional) (OMOP: date)',
  episode_end_datetime TIMESTAMP COMMENT 'End datetime (optional) (OMOP: datetime)',
  episode_parent_id BIGINT COMMENT 'FK -> episode.episode_id (optional) (OMOP: integer)',
  episode_number INT COMMENT 'Episode number (optional) (OMOP: integer)',
  episode_object_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Object concept) (OMOP: integer)',
  episode_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  episode_source_value STRING COMMENT 'Source value (optional) (OMOP: varchar(50))',
  episode_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 EPISODE table';
```
