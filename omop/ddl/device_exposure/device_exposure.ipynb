```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.device_exposure (
  device_exposure_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  device_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Device) (OMOP: integer)',
  device_exposure_start_date DATE NOT NULL COMMENT 'Start date (OMOP: date)',
  device_exposure_start_datetime TIMESTAMP COMMENT 'Start datetime (optional) (OMOP: datetime)',
  device_exposure_end_date DATE COMMENT 'End date (optional) (OMOP: date)',
  device_exposure_end_datetime TIMESTAMP COMMENT 'End datetime (optional) (OMOP: datetime)',
  device_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  unique_device_id STRING COMMENT 'UDI (optional) (OMOP: varchar(50))',
  quantity INT COMMENT 'Quantity (optional) (OMOP: integer)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  device_source_value STRING COMMENT 'Source device value (OMOP: varchar(50))',
  device_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 DEVICE_EXPOSURE table';
```
