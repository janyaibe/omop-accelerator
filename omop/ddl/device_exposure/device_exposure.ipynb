```python
%sql
CREATE OR REPLACE TABLE _exponent.omop.device_exposure (
    device_exposure_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
    person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
    device_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Device) (OMOP: integer)',
    device_exposure_start_date DATE NOT NULL COMMENT 'Device exposure start date (OMOP: date)',
    device_exposure_start_datetime TIMESTAMP COMMENT 'Device exposure start datetime (optional) (OMOP: datetime)',
    device_exposure_end_date DATE COMMENT 'Device exposure end date (optional) (OMOP: date)',
    device_exposure_end_datetime TIMESTAMP COMMENT 'Device exposure end datetime (optional) (OMOP: datetime)',
    device_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
    unique_device_id STRING COMMENT 'Unique device identifier (optional) (OMOP: varchar(255))',
    production_id STRING COMMENT 'Production identifier / lot / serial (optional) (OMOP: varchar(255))',
    quantity INT COMMENT 'Quantity of devices (optional) (OMOP: integer)',
    provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
    visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
    visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
    device_source_value STRING COMMENT 'Source device value (OMOP: varchar(50))',
    device_source_concept_id INT COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
    unit_concept_id INT COMMENT 'FK -> concept.concept_id (Unit) (OMOP: integer)',
    unit_source_value STRING COMMENT 'Source unit value (OMOP: varchar(50))',
    unit_source_concept_id INT COMMENT 'FK -> concept.concept_id (Source unit concept) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 DEVICE_EXPOSURE table';
```
