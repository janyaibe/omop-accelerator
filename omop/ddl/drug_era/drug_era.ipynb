```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.drug_era (
  drug_era_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  drug_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Drug) (OMOP: integer)',
  drug_era_start_date DATE NOT NULL COMMENT 'Start date (OMOP: date)',
  drug_era_end_date DATE NOT NULL COMMENT 'End date (OMOP: date)',
  drug_exposure_count INT COMMENT 'Count of contributing drug_exposure rows (OMOP: integer)',
  gap_days INT COMMENT 'Gap days (optional) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 DRUG_ERA table';
```
