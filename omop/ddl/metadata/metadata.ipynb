```python
%sql
CREATE OR REPLACE TABLE _exponent.omop.metadata (
    metadata_id INT NOT NULL COMMENT 'PK (OMOP: integer)',
    metadata_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Metadata concept)',
    metadata_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Metadata type)',
    name STRING NOT NULL COMMENT 'Metadata name (OMOP: varchar(250))',
    value_as_string STRING COMMENT 'Value as string (OMOP: varchar(250))',
    value_as_concept_id INT COMMENT 'FK -> concept.concept_id (Value concept)',
    value_as_number DOUBLE COMMENT 'Value as number (OMOP: float)',
    metadata_date DATE COMMENT 'Metadata date (OMOP: date)',
    metadata_datetime TIMESTAMP COMMENT 'Metadata datetime (OMOP: datetime)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 METADATA table';
```
