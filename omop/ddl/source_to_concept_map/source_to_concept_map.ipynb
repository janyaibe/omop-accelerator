```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.source_to_concept_map (
  source_code STRING NOT NULL COMMENT 'Source code (OMOP: varchar(50))',
  source_concept_id INT COMMENT 'FK -> concept.concept_id (optional) (OMOP: integer)',
  source_vocabulary_id STRING NOT NULL COMMENT 'Source vocabulary id (OMOP: varchar(20))',
  source_code_description STRING COMMENT 'Description (optional) (OMOP: varchar(255))',
  target_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (target) (OMOP: integer)',
  target_vocabulary_id STRING NOT NULL COMMENT 'Target vocabulary id (OMOP: varchar(20))',
  valid_start_date DATE NOT NULL COMMENT 'Valid start (OMOP: date)',
  valid_end_date DATE NOT NULL COMMENT 'Valid end (OMOP: date)',
  invalid_reason STRING COMMENT 'Invalid reason (optional) (OMOP: varchar(1))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 SOURCE_TO_CONCEPT_MAP table';
```
