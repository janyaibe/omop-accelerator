```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.concept (
  concept_id INT NOT NULL COMMENT 'PK. Unique concept identifier (OMOP: integer)',
  concept_name STRING NOT NULL COMMENT 'Concept name (OMOP: varchar(255))',
  domain_id STRING NOT NULL COMMENT 'FK -> domain.domain_id (OMOP: varchar(20))',
  vocabulary_id STRING NOT NULL COMMENT 'FK -> vocabulary.vocabulary_id (OMOP: varchar(20))',
  concept_class_id STRING NOT NULL COMMENT 'FK -> concept_class.concept_class_id (OMOP: varchar(20))',
  standard_concept STRING COMMENT 'Standard concept flag (S/C) (optional) (OMOP: varchar(1))',
  concept_code STRING NOT NULL COMMENT 'Source code (OMOP: varchar(50))',
  valid_start_date DATE NOT NULL COMMENT 'Valid start date (OMOP: date)',
  valid_end_date DATE NOT NULL COMMENT 'Valid end date (OMOP: date)',
  invalid_reason STRING COMMENT 'Invalid reason (D/U) (optional) (OMOP: varchar(1))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 CONCEPT table';
```
