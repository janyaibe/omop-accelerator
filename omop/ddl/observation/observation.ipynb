```python
%sql
CREATE OR REPLACE TABLE _exponent.omop.observation (
  observation_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  observation_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Observation) (OMOP: integer)',
  observation_date DATE NOT NULL COMMENT 'Observation date (OMOP: date)',
  observation_datetime TIMESTAMP COMMENT 'Observation datetime (optional) (OMOP: datetime)',
  observation_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  value_as_number DOUBLE COMMENT 'Numeric value (optional) (OMOP: numeric)',
  value_as_string STRING COMMENT 'String value (optional) (OMOP: varchar(60))',
  value_as_concept_id INT COMMENT 'FK -> concept.concept_id (Coded value) (OMOP: integer)',
  qualifier_concept_id INT COMMENT 'FK -> concept.concept_id (Qualifier) (OMOP: integer)',
  unit_concept_id INT COMMENT 'FK -> concept.concept_id (Unit) (OMOP: integer)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  observation_source_value STRING COMMENT 'Source observation value (OMOP: varchar(50))',
  observation_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
  unit_source_value STRING COMMENT 'Source unit (OMOP: varchar(50))',
  qualifier_source_value STRING COMMENT 'Source qualifier (optional) (OMOP: varchar(50))',
  value_source_value STRING COMMENT 'Source value (optional) (OMOP: varchar(50))',
  observation_event_id BIGINT COMMENT 'ID of the linked event record (OMOP: integer)',
  obs_event_field_concept_id INT COMMENT 'FK -> concept.concept_id indicating the field of the linked event (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 OBSERVATION table';
```
