```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.death (
  person_id BIGINT NOT NULL COMMENT 'PK + FK -> person.person_id (OMOP: integer)',
  death_date DATE NOT NULL COMMENT 'Date of death (OMOP: date)',
  death_datetime TIMESTAMP COMMENT 'Datetime of death (optional) (OMOP: datetime)',
  death_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  cause_concept_id INT COMMENT 'FK -> concept.concept_id (Cause) (OMOP: integer)',
  cause_source_value STRING COMMENT 'Source cause (OMOP: varchar(50))',
  cause_source_concept_id INT COMMENT 'FK -> concept.concept_id (Cause source concept) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 DEATH table';
```
