```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.specimen (
  specimen_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  specimen_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Specimen) (OMOP: integer)',
  specimen_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  specimen_date DATE NOT NULL COMMENT 'Specimen date (OMOP: date)',
  specimen_datetime TIMESTAMP COMMENT 'Specimen datetime (optional) (OMOP: datetime)',
  quantity DOUBLE COMMENT 'Quantity (optional) (OMOP: numeric)',
  unit_concept_id INT COMMENT 'FK -> concept.concept_id (Unit) (OMOP: integer)',
  anatomic_site_concept_id INT COMMENT 'FK -> concept.concept_id (Anatomic site) (OMOP: integer)',
  disease_status_concept_id INT COMMENT 'FK -> concept.concept_id (Disease status) (OMOP: integer)',
  specimen_source_id STRING COMMENT 'Source specimen id (optional) (OMOP: varchar(50))',
  specimen_source_value STRING NOT NULL COMMENT 'Source specimen value (OMOP: varchar(50))',
  unit_source_value STRING COMMENT 'Source unit (OMOP: varchar(50))',
  anatomic_site_source_value STRING COMMENT 'Source anatomic site (OMOP: varchar(50))',
  disease_status_source_value STRING COMMENT 'Source disease status (OMOP: varchar(50))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 SPECIMEN table';
```
