```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.cohort_attribute (
  cohort_definition_id INT NOT NULL COMMENT 'FK -> cohort_definition.cohort_definition_id (OMOP: integer)',
  subject_id BIGINT NOT NULL COMMENT 'Subject id (OMOP: integer)',
  cohort_start_date DATE NOT NULL COMMENT 'Cohort start date (OMOP: date)',
  cohort_end_date DATE NOT NULL COMMENT 'Cohort end date (OMOP: date)',
  attribute_definition_id INT NOT NULL COMMENT 'FK -> attribute_definition.attribute_definition_id (OMOP: integer)',
  value_as_number DOUBLE COMMENT 'Numeric value (optional) (OMOP: numeric)',
  value_as_concept_id INT COMMENT 'FK -> concept.concept_id (Coded value, optional) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 COHORT_ATTRIBUTE table';
```
