```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.person (
  person_id BIGINT NOT NULL COMMENT 'PK. Unique identifier for each person (OMOP: integer)',
  gender_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Gender) (OMOP: integer)',
  year_of_birth INT NOT NULL COMMENT 'Year of birth (OMOP: integer)',
  month_of_birth INT COMMENT 'Month of birth (optional) (OMOP: integer)',
  day_of_birth INT COMMENT 'Day of birth (optional) (OMOP: integer)',
  birth_datetime TIMESTAMP COMMENT 'Birth datetime (optional) (OMOP: datetime)',
  race_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Race) (OMOP: integer)',
  ethnicity_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Ethnicity) (OMOP: integer)',
  location_id BIGINT COMMENT 'FK -> location.location_id (OMOP: integer)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  care_site_id BIGINT COMMENT 'FK -> care_site.care_site_id (OMOP: integer)',
  person_source_value STRING COMMENT 'Source person identifier (OMOP: varchar(50))',
  gender_source_value STRING COMMENT 'Source gender value (OMOP: varchar(50))',
  gender_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Gender source concept) (OMOP: integer)',
  race_source_value STRING COMMENT 'Source race value (OMOP: varchar(50))',
  race_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Race source concept) (OMOP: integer)',
  ethnicity_source_value STRING COMMENT 'Source ethnicity value (OMOP: varchar(50))',
  ethnicity_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Ethnicity source concept) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 PERSON table';
```
