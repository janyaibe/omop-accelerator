```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.fact_relationship (
  domain_concept_id_1 INT NOT NULL COMMENT 'FK -> concept.concept_id (Domain of fact 1) (OMOP: integer)',
  fact_id_1 BIGINT NOT NULL COMMENT 'Identifier of fact 1 in its table (OMOP: integer)',
  domain_concept_id_2 INT NOT NULL COMMENT 'FK -> concept.concept_id (Domain of fact 2) (OMOP: integer)',
  fact_id_2 BIGINT NOT NULL COMMENT 'Identifier of fact 2 in its table (OMOP: integer)',
  relationship_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Relationship) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 FACT_RELATIONSHIP table';
```
