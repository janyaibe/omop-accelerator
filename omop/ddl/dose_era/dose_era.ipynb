```python
%sql
CREATE OR REPLACE TABLE _exponent.omop.dose_era (
  dose_era_id BIGINT NOT NULL
    GENERATED BY DEFAULT AS IDENTITY (START WITH 1 INCREMENT BY 1)
    COMMENT 'PK (OMOP: integer)',

  person_id BIGINT NOT NULL
    COMMENT 'FK -> person.person_id (OMOP: integer)',

  drug_concept_id INT NOT NULL
    COMMENT 'FK -> concept.concept_id (Drug) (OMOP: integer)',

  unit_concept_id INT NOT NULL
    COMMENT 'FK -> concept.concept_id (Unit) (OMOP: integer)',

  dose_value DOUBLE
    COMMENT 'Dose value (optional) (OMOP: numeric)',

  dose_era_start_date DATE NOT NULL
    COMMENT 'Start date (OMOP: date)',

  dose_era_end_date DATE NOT NULL
    COMMENT 'End date (OMOP: date)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 DOSE_ERA table';
```
