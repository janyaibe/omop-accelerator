```python
%sql
CREATE OR REPLACE TABLE _exponent.omop.concept_ancestor (
  ancestor_concept_id INT NOT NULL COMMENT 'Ancestor concept_id',
  descendant_concept_id INT NOT NULL COMMENT 'Descendant concept_id',
  min_levels_of_separation INT NOT NULL COMMENT 'Minimum number of hierarchical steps',
  max_levels_of_separation INT NOT NULL COMMENT 'Maximum number of hierarchical steps'
)
USING DELTA
COMMENT 'OMOP vocabulary table containing hierarchical ancestor-descendant relationships between concepts';
```
