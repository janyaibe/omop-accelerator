```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.provider (
  provider_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  provider_name STRING COMMENT 'Provider name (OMOP: varchar(255))',
  npi STRING COMMENT 'NPI (optional) (OMOP: varchar(20))',
  dea STRING COMMENT 'DEA (optional) (OMOP: varchar(20))',
  specialty_concept_id INT COMMENT 'FK -> concept.concept_id (Specialty) (OMOP: integer)',
  care_site_id BIGINT COMMENT 'FK -> care_site.care_site_id (OMOP: integer)',
  year_of_birth INT COMMENT 'Year of birth (optional) (OMOP: integer)',
  gender_concept_id INT COMMENT 'FK -> concept.concept_id (Gender) (OMOP: integer)',
  provider_source_value STRING COMMENT 'Source provider value (OMOP: varchar(50))',
  specialty_source_value STRING COMMENT 'Source specialty (OMOP: varchar(50))',
  specialty_source_concept_id INT COMMENT 'FK -> concept.concept_id (Specialty source concept) (OMOP: integer)',
  gender_source_value STRING COMMENT 'Source gender (OMOP: varchar(50))',
  gender_source_concept_id INT COMMENT 'FK -> concept.concept_id (Gender source concept) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 PROVIDER table';
```
