```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.payer_plan_period (
  payer_plan_period_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  payer_plan_period_start_date DATE NOT NULL COMMENT 'Start date (OMOP: date)',
  payer_plan_period_end_date DATE NOT NULL COMMENT 'End date (OMOP: date)',
  payer_concept_id INT COMMENT 'FK -> concept.concept_id (Payer) (OMOP: integer)',
  payer_source_value STRING COMMENT 'Source payer (OMOP: varchar(50))',
  payer_source_concept_id INT COMMENT 'FK -> concept.concept_id (Payer source concept) (OMOP: integer)',
  plan_concept_id INT COMMENT 'FK -> concept.concept_id (Plan) (OMOP: integer)',
  plan_source_value STRING COMMENT 'Source plan (OMOP: varchar(50))',
  plan_source_concept_id INT COMMENT 'FK -> concept.concept_id (Plan source concept) (OMOP: integer)',
  sponsor_concept_id INT COMMENT 'FK -> concept.concept_id (Sponsor) (OMOP: integer)',
  sponsor_source_value STRING COMMENT 'Source sponsor (OMOP: varchar(50))',
  sponsor_source_concept_id INT COMMENT 'FK -> concept.concept_id (Sponsor source concept) (OMOP: integer)',
  family_source_value STRING COMMENT 'Source family (optional) (OMOP: varchar(50))',
  stop_reason_concept_id INT COMMENT 'FK -> concept.concept_id (Stop reason) (OMOP: integer)',
  stop_reason_source_value STRING COMMENT 'Source stop reason (OMOP: varchar(50))',
  stop_reason_source_concept_id INT COMMENT 'FK -> concept.concept_id (Stop reason source concept) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 PAYER_PLAN_PERIOD table';
```
