```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.condition_occurrence (
  condition_occurrence_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  condition_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Condition) (OMOP: integer)',
  condition_start_date DATE NOT NULL COMMENT 'Start date (OMOP: date)',
  condition_start_datetime TIMESTAMP COMMENT 'Start datetime (optional) (OMOP: datetime)',
  condition_end_date DATE COMMENT 'End date (optional) (OMOP: date)',
  condition_end_datetime TIMESTAMP COMMENT 'End datetime (optional) (OMOP: datetime)',
  condition_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  condition_status_concept_id INT COMMENT 'FK -> concept.concept_id (Status) (OMOP: integer)',
  stop_reason STRING COMMENT 'Stop reason (optional) (OMOP: varchar(20))',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  condition_source_value STRING COMMENT 'Source condition code/value (OMOP: varchar(50))',
  condition_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
  condition_status_source_value STRING COMMENT 'Source status value (OMOP: varchar(50))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 CONDITION_OCCURRENCE table';
```
