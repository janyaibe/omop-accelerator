```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.visit_occurrence (
  visit_occurrence_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  visit_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Visit) (OMOP: integer)',
  visit_start_date DATE NOT NULL COMMENT 'Visit start date (OMOP: date)',
  visit_start_datetime TIMESTAMP COMMENT 'Visit start datetime (optional) (OMOP: datetime)',
  visit_end_date DATE NOT NULL COMMENT 'Visit end date (OMOP: date)',
  visit_end_datetime TIMESTAMP COMMENT 'Visit end datetime (optional) (OMOP: datetime)',
  visit_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Visit type) (OMOP: integer)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  care_site_id BIGINT COMMENT 'FK -> care_site.care_site_id (OMOP: integer)',
  visit_source_value STRING COMMENT 'Source visit value (OMOP: varchar(50))',
  visit_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Visit source concept) (OMOP: integer)',
  admitted_from_concept_id INT COMMENT 'FK -> concept.concept_id (Admit source) (OMOP: integer)',
  admitted_from_source_value STRING COMMENT 'Admit source value (OMOP: varchar(50))',
  discharged_to_concept_id INT COMMENT 'FK -> concept.concept_id (Discharge to) (OMOP: integer)',
  discharged_to_source_value STRING COMMENT 'Discharge destination value (OMOP: varchar(50))',
  preceding_visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (optional) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 VISIT_OCCURRENCE table';
```
