```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.cohort (
  cohort_definition_id INT NOT NULL COMMENT 'FK -> cohort_definition.cohort_definition_id (OMOP: integer)',
  subject_id BIGINT NOT NULL COMMENT 'Subject id (OMOP: integer)',
  cohort_start_date DATE NOT NULL COMMENT 'Start date (OMOP: date)',
  cohort_end_date DATE NOT NULL COMMENT 'End date (OMOP: date)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 COHORT table';
```
