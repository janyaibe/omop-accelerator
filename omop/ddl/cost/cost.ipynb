```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.cost (
  cost_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  cost_event_id BIGINT NOT NULL COMMENT 'Identifier of event being costed (OMOP: integer)',
  cost_domain_id STRING NOT NULL COMMENT 'Domain id of the event (OMOP: varchar(20))',
  cost_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  currency_concept_id INT COMMENT 'FK -> concept.concept_id (Currency) (OMOP: integer)',
  total_charge DOUBLE COMMENT 'Total charge (optional) (OMOP: numeric)',
  total_cost DOUBLE COMMENT 'Total cost (optional) (OMOP: numeric)',
  total_paid DOUBLE COMMENT 'Total paid (optional) (OMOP: numeric)',
  paid_by_payer DOUBLE COMMENT 'Paid by payer (optional) (OMOP: numeric)',
  paid_by_patient DOUBLE COMMENT 'Paid by patient (optional) (OMOP: numeric)',
  paid_patient_copay DOUBLE COMMENT 'Copay (optional) (OMOP: numeric)',
  paid_patient_coinsurance DOUBLE COMMENT 'Coinsurance (optional) (OMOP: numeric)',
  paid_patient_deductible DOUBLE COMMENT 'Deductible (optional) (OMOP: numeric)',
  paid_by_primary DOUBLE COMMENT 'Paid by primary (optional) (OMOP: numeric)',
  paid_ingredient_cost DOUBLE COMMENT 'Ingredient cost (optional) (OMOP: numeric)',
  paid_dispensing_fee DOUBLE COMMENT 'Dispensing fee (optional) (OMOP: numeric)',
  payer_plan_period_id BIGINT COMMENT 'FK -> payer_plan_period.payer_plan_period_id (optional) (OMOP: integer)',
  amount_allowed DOUBLE COMMENT 'Allowed amount (optional) (OMOP: numeric)',
  revenue_code_concept_id INT COMMENT 'FK -> concept.concept_id (Revenue code) (OMOP: integer)',
  revenue_code_source_value STRING COMMENT 'Source revenue code (OMOP: varchar(50))',
  drg_concept_id INT COMMENT 'FK -> concept.concept_id (DRG) (OMOP: integer)',
  drg_source_value STRING COMMENT 'Source DRG (OMOP: varchar(3))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 COST table';
```
