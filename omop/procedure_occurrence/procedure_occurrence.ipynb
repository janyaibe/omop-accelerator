```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.procedure_occurrence (
  procedure_occurrence_id BIGINT NOT NULL COMMENT 'PK (OMOP: integer)',
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id (OMOP: integer)',
  procedure_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Procedure) (OMOP: integer)',
  procedure_date DATE NOT NULL COMMENT 'Procedure date (OMOP: date)',
  procedure_datetime TIMESTAMP COMMENT 'Procedure datetime (optional) (OMOP: datetime)',
  procedure_end_date DATE COMMENT 'Procedure end date (optional) (OMOP: date)',
  procedure_end_datetime TIMESTAMP COMMENT 'Procedure end datetime (optional) (OMOP: datetime)',
  procedure_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Type) (OMOP: integer)',
  modifier_concept_id INT COMMENT 'FK -> concept.concept_id (Modifier) (OMOP: integer)',
  quantity INT COMMENT 'Quantity (optional) (OMOP: integer)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id (OMOP: integer)',
  visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id (OMOP: integer)',
  visit_detail_id BIGINT COMMENT 'FK -> visit_detail.visit_detail_id (OMOP: integer)',
  procedure_source_value STRING COMMENT 'Source procedure value (OMOP: varchar(50))',
  procedure_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Source concept) (OMOP: integer)',
  modifier_source_value STRING COMMENT 'Source modifier (optional) (OMOP: varchar(50))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 PROCEDURE_OCCURRENCE table';
```
