```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.vocabulary (
  vocabulary_id STRING NOT NULL COMMENT 'Unique identifier for the vocabulary (OMOP: varchar(20))',
  vocabulary_name STRING NOT NULL COMMENT 'Full name of the vocabulary (OMOP: varchar(255))',
  vocabulary_reference STRING COMMENT 'Reference URL or citation (OMOP: varchar(255))',
  vocabulary_version STRING COMMENT 'Version identifier (OMOP: varchar(255))',
  vocabulary_concept_id INT NOT NULL COMMENT 'Concept representing the vocabulary (FK -> concept.concept_id) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 VOCABULARY table';
```
