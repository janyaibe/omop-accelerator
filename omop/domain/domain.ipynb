```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.domain (
  domain_id STRING NOT NULL COMMENT 'Unique key for each domain (OMOP: varchar(20))',
  domain_name STRING NOT NULL COMMENT 'Domain name (OMOP: varchar(255))',
  domain_concept_id INT NOT NULL COMMENT 'Concept representing the Domain (FK -> concept.concept_id) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 DOMAIN table';
```
