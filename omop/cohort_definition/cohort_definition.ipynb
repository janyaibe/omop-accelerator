```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.cohort_definition (
  cohort_definition_id INT NOT NULL COMMENT 'PK (OMOP: integer)',
  cohort_definition_name STRING NOT NULL COMMENT 'Name (OMOP: varchar(255))',
  cohort_definition_description STRING COMMENT 'Description (optional) (OMOP: varchar(1000))',
  definition_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Definition type) (OMOP: integer)',
  cohort_definition_syntax STRING COMMENT 'Cohort definition expression/syntax (optional) (OMOP: varchar(MAX))',
  subject_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Subject) (OMOP: integer)'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 COHORT_DEFINITION table';
```
