```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop.drug_strength (
  drug_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Drug) (OMOP: integer)',
  ingredient_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Ingredient) (OMOP: integer)',
  amount_value DOUBLE COMMENT 'Amount value (optional) (OMOP: numeric)',
  amount_unit_concept_id INT COMMENT 'FK -> concept.concept_id (Amount unit, optional) (OMOP: integer)',
  numerator_value DOUBLE COMMENT 'Numerator value (optional) (OMOP: numeric)',
  numerator_unit_concept_id INT COMMENT 'FK -> concept.concept_id (Numerator unit, optional) (OMOP: integer)',
  denominator_value DOUBLE COMMENT 'Denominator value (optional) (OMOP: numeric)',
  denominator_unit_concept_id INT COMMENT 'FK -> concept.concept_id (Denominator unit, optional) (OMOP: integer)',
  box_size INT COMMENT 'Box size (optional) (OMOP: integer)',
  valid_start_date DATE NOT NULL COMMENT 'Valid start (OMOP: date)',
  valid_end_date DATE NOT NULL COMMENT 'Valid end (OMOP: date)',
  invalid_reason STRING COMMENT 'Invalid reason (optional) (OMOP: varchar(1))'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 DRUG_STRENGTH table';
```
