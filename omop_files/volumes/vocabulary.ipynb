```python
from pyspark.sql.functions import col
```

```python


vocabulary_df = (
    spark.read
        .option("header", True)
        .option("sep", "\t")
        .csv("/Volumes/_exponent/omop_files/vocabulary/VOCABULARY.csv")
        .withColumn("vocabulary_concept_id", col("vocabulary_concept_id").cast("int"))
)

vocabulary_df.write.format("delta").mode("overwrite").saveAsTable("_exponent.omop.vocabulary")
```
