```python
from pyspark.sql.functions import col, to_date

concept_df = spark.read.option("header", True).option("sep", "\t").csv("/Volumes/_exponent/omop_files/vocabulary/CONCEPT.csv")

concept_df = (
    concept_df
    .select(
        col("concept_id").cast("int").alias("concept_id"),
        col("concept_name").cast("string").alias("concept_name"),
        col("domain_id").cast("string").alias("domain_id"),
        col("vocabulary_id").cast("string").alias("vocabulary_id"),
        col("concept_class_id").cast("string").alias("concept_class_id"),
        col("standard_concept").cast("string").alias("standard_concept"),
        col("concept_code").cast("string").alias("concept_code"),
        to_date(col("valid_start_date"), "yyyyMMdd").alias("valid_start_date"),
        to_date(col("valid_end_date"), "yyyyMMdd").alias("valid_end_date"),
        col("invalid_reason").cast("string").alias("invalid_reason")
    )
)

concept_df.write.format("delta").mode("overwrite").saveAsTable("_exponent.omop.concept")
```
