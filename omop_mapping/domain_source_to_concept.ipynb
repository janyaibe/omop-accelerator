```python
%sql
-- =========================================================
-- Domain Source → OMOP Concept Mapping
-- Purpose:
--   Authoritative mapping table used during OMOP ETL to map
--   source-system domain values (e.g., Race, Gender, Ethnicity)
--   to OMOP standard concept_ids.
--
-- Notes:
--   • This table supports multiple source systems.
--   • Source vocabularies are NOT loaded into OMOP.CONCEPT.
--   • source_id/source_value come from source dictionary tables.
--   • omop_concept_id must reference OMOP.CONCEPT.CONCEPT_ID.
-- =========================================================

CREATE TABLE IF NOT EXISTS _exponent.omop_mapping.domain_source_to_concept (
  
  source_system STRING
    COMMENT 'Source EHR or system name (e.g., allscripts_tw, epic)',

  source_table STRING
    COMMENT 'Source dictionary or lookup table name (e.g., dbo_race_de)',

  source_field STRING
    COMMENT 'Field in source_table providing the decoded value (e.g., EntryName)',

  domain_id STRING
    COMMENT 'OMOP domain this mapping applies to (Race, Ethnicity, Gender, Route, Unit, etc.)',

  source_id INT
    COMMENT 'Stable identifier from the source dictionary table (e.g., dbo_race_de.id)',

  source_value STRING
    COMMENT 'Human-readable decoded source value (e.g., Black or African American)',

  omop_concept_id INT
    COMMENT 'OMOP standard concept_id mapped from the source value (FK to OMOP.CONCEPT.CONCEPT_ID)',

  omop_concept_name STRING
    COMMENT 'OMOP standard concept_name for readability and validation',

  active_flag INT
    COMMENT 'Indicates whether this mapping is active (1 = active, 0 = inactive)',

  last_update_tsp TIMESTAMP
    COMMENT 'Timestamp when the mapping row was last updated'
)
USING DELTA
COMMENT 'Maps source-system domain dictionary values to OMOP standard concepts for use during OMOP transformations';
```
