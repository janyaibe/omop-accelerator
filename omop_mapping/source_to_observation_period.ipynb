```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop_mapping.source_to_observation_period (
    source_system                       STRING  COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    observation_period_source_value     STRING  COMMENT 'Prefixed source observation_period identifier (e.g., allscripts_tw | 12345)',
    observation_period_id               BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP observation_period_id',
    active_flag                         BOOLEAN COMMENT 'TRUE if this mapping is currently active',
    created_tsp                         TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp                        TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    CONSTRAINT pk_source_to_observation_period
        PRIMARY KEY (observation_period_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system observation_period identifiers to canonical OMOP observation_period_id';
```
