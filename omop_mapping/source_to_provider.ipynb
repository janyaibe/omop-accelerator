```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_mapping.source_to_provider (
    source_system            STRING    COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    provider_source_value    STRING    COMMENT 'Prefixed source provider identifier (e.g., allscripts_tw|98765)',
    provider_id              BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP provider_id',
    active_flag              BOOLEAN   COMMENT '1 if this mapping is currently active',
    created_tsp              TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp             TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    merge_id                 BIGINT    COMMENT 'Previous provider_id if this row was reassigned during a merge',
    merge_reason             STRING    COMMENT 'Reason for reassignment (manual merge, algorithmic match, etc.)',
    CONSTRAINT pk_source_to_provider
        PRIMARY KEY (provider_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system provider identifiers to canonical OMOP provider_id';
```
