```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_mapping.source_to_person (
    source_system           STRING  COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    person_source_value     STRING  COMMENT 'Prefixed source person identifier (e.g., allscripts_tw|12345)',
    person_id               BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP person_id',
    active_flag             BOOLEAN COMMENT '1 if this mapping is currently active',
    created_tsp             TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp            TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    merge_id                BIGINT  COMMENT 'Previous person_id if this row was reassigned during a merge',
    merge_reason            STRING  COMMENT 'Reason for reassignment (manual merge, algorithmic match, etc.)',
    CONSTRAINT pk_source_to_person
        PRIMARY KEY (person_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system person identifiers to canonical OMOP person_id';
```
