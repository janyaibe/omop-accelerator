```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_mapping.source_to_device_exposure (
    source_system                   STRING  COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    device_exposure_source_value    STRING  COMMENT 'Prefixed source device_exposure identifier (ex: CONCAT_WS(CHR(31), ...))',
    device_exposure_id              BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP device_exposure_id',
    active_flag                     BOOLEAN COMMENT '1 if this mapping is currently active',
    created_tsp                     TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp                    TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    merge_id                        BIGINT  COMMENT 'Previous device_exposure_id if this row was reassigned during a merge',
    merge_reason                    STRING  COMMENT 'Reason for reassignment (manual merge, algorithmic match, etc.)',
    CONSTRAINT pk_source_to_device_exposure
        PRIMARY KEY (device_exposure_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system device_exposure identifiers to canonical OMOP device_exposure_id';
```
