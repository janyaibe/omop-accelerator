```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop_mapping.source_to_drug_exposure (
    source_system                   STRING  COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    drug_exposure_source_value      STRING  COMMENT 'Prefixed source drug_exposure identifier (e.g., allscripts_tw | 12345)',
    drug_exposure_id                BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP drug_exposure_id',
    active_flag                     BOOLEAN COMMENT 'TRUE if this mapping is currently active',
    created_tsp                     TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp                    TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    CONSTRAINT pk_source_to_drug_exposure
        PRIMARY KEY (drug_exposure_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system drug_exposure identifiers to canonical OMOP drug_exposure_id';
```
