```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_mapping.source_to_location (
    source_system           STRING  COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    location_source_value     STRING  COMMENT 'Prefixed source location identifier (ex: CONCAT_WS(CHR(31), ...))',
    location_id               BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP location_id',
    active_flag             BOOLEAN COMMENT '1 if this mapping is currently active',
    created_tsp             TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp            TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    merge_id                BIGINT  COMMENT 'Previous location_id if this row was reassigned during a merge',
    merge_reason            STRING  COMMENT 'Reason for reassignment (manual merge, algorithmic match, etc.)',
    CONSTRAINT pk_source_to_location
        PRIMARY KEY (location_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system location identifiers to canonical OMOP location_id';
```
