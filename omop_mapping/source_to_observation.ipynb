```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_mapping.source_to_observation (
    source_system                   STRING  COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    observation_source_value        STRING  COMMENT 'Prefixed source observation identifier (e.g., allscripts_tw|209987094800063)',
    observation_id                  BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP observation_id',
    person_id                       BIGINT  COMMENT 'FK -> person.person_id - links observation to person',
    active_flag                     BOOLEAN COMMENT '1 if this mapping is currently active',
    created_tsp                     TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp                    TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    merge_id                        BIGINT  COMMENT 'Previous observation_id if this row was reassigned during a merge',
    merge_reason                    STRING  COMMENT 'Reason for reassignment (manual merge, algorithmic match, etc.)',
    CONSTRAINT pk_source_to_observation
        PRIMARY KEY (observation_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system observation identifiers to canonical OMOP observation_id';
```
