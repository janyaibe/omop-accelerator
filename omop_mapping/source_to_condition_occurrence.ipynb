```python
%sql
CREATE TABLE IF NOT EXISTS _exponent.omop_mapping.source_to_condition_occurrence (
    source_system                         STRING  COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    condition_occurrence_source_value     STRING  COMMENT 'Prefixed source condition_occurrence identifier (e.g., allscripts_tw | 12345)',
    condition_occurrence_id               BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP condition_occurrence_id',
    active_flag                           BOOLEAN COMMENT 'TRUE if this mapping is currently active',
    created_tsp                           TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp                          TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    CONSTRAINT pk_source_to_condition_occurrence
        PRIMARY KEY (condition_occurrence_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system condition_occurrence identifiers to canonical OMOP condition_occurrence_id';
```
