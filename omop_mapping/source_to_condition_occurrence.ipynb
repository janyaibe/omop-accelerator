```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_mapping.source_to_condition_occurrence (
    source_system                     STRING    COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    condition_occurrence_source_value  STRING    NOT NULL COMMENT 'Prefixed source condition occurrence identifier (e.g., CONCAT_WS(CHR(31), ...))',
    condition_occurrence_id            BIGINT    GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP condition_occurrence_id',
    active_flag                        BOOLEAN   COMMENT 'TRUE if this mapping is currently active',
    created_tsp                        TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp                       TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    merge_id                           BIGINT    COMMENT 'Previous condition_occurrence_id if this row was reassigned during a merge',
    merge_reason                       STRING    COMMENT 'Reason for reassignment (manual merge, algorithmic match, etc.)',
    CONSTRAINT pk_source_to_condition_occurrence
        PRIMARY KEY (condition_occurrence_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system condition occurrence identifiers to canonical OMOP condition_occurrence_id';
```
