```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_mapping.source_to_visitor_occurrence (
    source_system                   STRING  COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    visit_occurrence_source_value   STRING  COMMENT 'Prefixed source visit/encounter identifier (e.g., allscripts_tw|179663788)',
    visit_occurrence_id             BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP visit_occurrence_id',
    person_id                       BIGINT  COMMENT 'FK -> person.person_id - links visit to person',
    active_flag                     BOOLEAN COMMENT '1 if this mapping is currently active',
    created_tsp                     TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp                    TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    merge_id                        BIGINT  COMMENT 'Previous visit_occurrence_id if this row was reassigned during a merge',
    merge_reason                    STRING  COMMENT 'Reason for reassignment (manual merge, algorithmic match, etc.)',
    CONSTRAINT pk_source_to_visitor_occurrence
        PRIMARY KEY (visit_occurrence_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system visit/encounter identifiers to canonical OMOP visit_occurrence_id';
```

```python
CREATE TABLE IF NOT EXISTS _exponent.omop_silver.visit_occurrence (
  person_id BIGINT NOT NULL COMMENT 'FK -> person.person_id',
  visit_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Visit)',
  visit_start_date DATE NOT NULL COMMENT 'Visit start date',
  visit_start_datetime TIMESTAMP COMMENT 'Visit start datetime',
  visit_end_date DATE NOT NULL COMMENT 'Visit end date',
  visit_end_datetime TIMESTAMP COMMENT 'Visit end datetime',
  visit_type_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id (Visit type)',
  provider_id BIGINT COMMENT 'FK -> provider.provider_id',
  care_site_id BIGINT COMMENT 'FK -> care_site.care_site_id',
  visit_source_value STRING COMMENT 'Source visit value',
  visit_source_concept_id INT NOT NULL COMMENT 'FK -> concept.concept_id',
  admitted_from_concept_id INT COMMENT 'FK -> concept.concept_id',
  admitted_from_source_value STRING COMMENT 'Admit source value',
  discharged_to_concept_id INT COMMENT 'FK -> concept.concept_id',
  discharged_to_source_value STRING COMMENT 'Discharge destination value',
  preceding_visit_occurrence_id BIGINT COMMENT 'FK -> visit_occurrence.visit_occurrence_id',
  source_system STRING COMMENT 'Source system identifier',
  last_mod_tsp TIMESTAMP COMMENT 'Last modified timestamp'
)
USING DELTA
COMMENT 'OMOP CDM v5.4 VISIT_OCCURRENCE silver layer';
```
