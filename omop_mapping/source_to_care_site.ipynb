```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_mapping.source_to_care_site (
  care_site_id BIGINT GENERATED BY DEFAULT AS IDENTITY (START WITH 1 INCREMENT BY 1)
    COMMENT 'Canonical OMOP care_site_id',
  source_system STRING NOT NULL COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
  care_site_source_value STRING NOT NULL COMMENT 'Prefixed source care_site identifier (e.g., CONCAT_WS(CHR(31), ...))',
  active_flag BOOLEAN NOT NULL COMMENT 'TRUE if this mapping is currently active',
  created_tsp TIMESTAMP NOT NULL COMMENT 'Timestamp when this mapping was first created',
  last_mod_tsp TIMESTAMP NOT NULL COMMENT 'Timestamp when this mapping was last updated',
  merge_id BIGINT COMMENT 'Previous care_site_id if this row was reassigned during a merge',
  merge_reason STRING COMMENT 'Reason for reassignment (manual merge, algorithmic match, etc.)',
  CONSTRAINT pk_source_to_care_site PRIMARY KEY (care_site_source_value)
)
USING DELTA
COMMENT 'Crosswalk from source care_site keys to OMOP care_site_id';
```
