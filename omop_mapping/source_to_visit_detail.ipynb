```python
%sql
CREATE OR REPLACE TABLE _exponent.omop_mapping.source_to_visit_detail (
    source_system                 STRING  COMMENT 'Source system identifier (e.g., allscripts_tw, epic)',
    visit_detail_source_value      STRING  COMMENT 'Prefixed source visit_detail identifier (ex: CONCAT_WS(CHR(31), ...))',
    visit_detail_id                BIGINT GENERATED ALWAYS AS IDENTITY COMMENT 'Canonical OMOP visit_detail_id',
    active_flag                    BOOLEAN COMMENT '1 if this mapping is currently active',
    created_tsp                    TIMESTAMP COMMENT 'Timestamp when this mapping was first created',
    last_mod_tsp                   TIMESTAMP COMMENT 'Timestamp when this mapping was last updated',
    merge_id                       BIGINT  COMMENT 'Previous visit_detail_id if this row was reassigned during a merge',
    merge_reason                   STRING  COMMENT 'Reason for reassignment (manual merge, algorithmic match, etc.)',
    CONSTRAINT pk_source_to_visit_detail
        PRIMARY KEY (visit_detail_source_value)
)
USING DELTA
COMMENT 'Crosswalk mapping source-system visit_detail identifiers to canonical OMOP visit_detail_id';
```
